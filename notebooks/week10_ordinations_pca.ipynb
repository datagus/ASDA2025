# Principal Component Analysis

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from pca import pca
from sklearn import datasets
from sklearn.decomposition import PCA
```

## Loading data

```python
wine_df = pd.read_csv("datasets/winequality-red.csv", sep=";")
wine_df["quality"] = wine_df["quality"].astype(str)
```

```python
wine_df["quality"].value_counts()
```

```python
wine_df
```

## Runnning PCA

```python
df = wine_df.copy()
y = df.pop("quality")
X = df
labels = X.columns

#scaling
scaler = StandardScaler()
X = scaler.fit_transform(X)

#initializing model
model = pca()

# Fit transform and include the column labels and row labels
results = model.fit_transform(X, col_labels=labels, row_labels = y)

# Scatter plot with loadings
model.biplot();
```

## Loadings

```python
model.results['topfeat']
```

```python
model.plot();
```
